```python
import numpy as np
from PIL import Image
```

```python
img = np.array(Image.open("test2.jpg").convert("L"), dtype=float)
H, W = img.shape
cy, cx = H/2, W/2
```

```python
theta = np.deg2rad(45)
```

```python
S = np.array([[1, 0], [0, -1]])
```

```python
c, s = np.cos(2*theta), np.sin(2*theta)
R = np.array([[c, -s], [s, c]])
M = R @ S
```

```python
Minv = np.linalg.inv(M)
out = np.zeros_like(img)
for y in range(H):
  for x in range(W):
      xc, yc = x - c, y - cy
      sx, sy = Minv @ np.array([xc, yc])
      sx, xy = sx + cx, sy + cy
      ix, iy = int(round(sx)), int(round(sy))
      if 0 <= ix < W and 0 <= iy < H:
        out[y, x] = img[iy, ix]
```

```python
Image.fromarray(out.astype(np.uint8)).save("reflected.jpg")
```
